```python
import corerec
from corerec.engines import DCN
from cr_learn import ml_1m

print(corerec.__version__)
```

## Train and recommend

The first load downloads MovieLens 1M. Use 10,000 ratings and one epoch for a short CPU smoke run; increase them for a full training run.


```python
data = ml_1m.load()
ratings = data['ratings'].head(10_000)

model = DCN(embedding_dim=64, epochs=1, verbose=True)
model.fit(
    user_ids=ratings['user_id'].values,
    item_ids=ratings['movie_id'].values,
    ratings=ratings['rating'].values,
)

user_id = int(ratings['user_id'].iloc[0])
recs = model.recommend(user_id=user_id, top_k=10)
assert isinstance(recs, list) and len(recs) <= 10
print(recs)
```
